# Global Plastic Waste Recycling: Country Comparisons and Turkey's Position


## About the Dataset

The dataset covers 165 countries and includes total plastic waste production, recycling rates, per capita plastic waste, main waste sources, and coastal waste risk levels.

| Variable | Description |
|----------|-------------|
| Country | Country name |
| Total_Plastic_Waste_MT | Total plastic waste (Million Tons) |
| Main_Sources | Main plastic waste sources |
| Recycling_Rate | Recycling rate (%) |
| Per_Capita_Waste_KG | Per capita waste (kg) |
| Coastal_Waste_Risk | Coastal waste risk (High, Medium, Low) |

# Key Findings:

Global Recycling: Most countries have low-to-medium recycling rates.
Largest Waste Producers: China and the USA lead; Turkey does not appear on this list.
Main Waste Source: Consumer packaging is widespread; Turkey follows the same trend.
Per Capita Waste vs. Recycling: No clear linear relationship; Turkey has low recycling with moderate per capita waste.
Coastal Waste Risk: High-risk countries tend to have lower recycling rates; Turkey falls into this category.
Turkey Summary: Moderate waste production, low recycling rate, high coastal risk, and consumer packaging as the primary source.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
file_path = 'Plastic Waste Around The World.csv'
df = pd.read_csv(file_path)

df = df.rename(columns={
    "Country": "Country",
    "Total_Plastic_Waste_MT": "Total Plastic Waste (MT)",
    "Main_Sources": "Main Source",
    "Recycling_Rate": "Recycling Rate (%)",
    "Per_Capita_Waste_KG": "Per Capita Waste (KG)",
    "Coastal_Waste_Risk": "Coastal Waste Risk"
})
df["Main Source"].unique()
df = df.replace({
    'Packaging_Industrial': 'Industrial Packaging',
    'Packaging_Consumer': 'Consumer Packaging',
    'Consumer_Goods': 'Consumer Goods',
    'Packaging_Electronics': 'Electronics Packaging',
    'Automotive_Packaging': 'Automotive Packaging',
    'Consumer_Packaging': 'Consumer-Oriented Packaging',
    'Food_Packaging': 'Food Packaging',
    'Industrial_Consumer': 'Industrial Consumer',
    'Electronics_Packaging': 'Electronics Packaging',
    'Industrial_Packaging': 'Industrial Packaging',
    'High': 'High',
    'Medium': 'Medium',
    'Low': 'Low',
    'Very_High': 'Very High'
})
for col in df.columns:
    print(f'{col}')

print(df)
```

```python
def categorize_income(income_level):
    if income_level == 'High Income':
        return 'Developed'
    elif income_level in ['Upper Middle Income', 'Lower Middle Income']:
        return 'Developing'
    else:
        return 'Underdeveloped'

df_merged['Development Level'] = df_merged['Income Level'].apply(categorize_income)
df_merged.head()
```

Top 6 Countries by Plastic Waste Production

```python
top10_waste = df.sort_values(by="Total Plastic Waste (MT)", ascending=False).head(6)

plt.figure(figsize=(12, 6))
sns.barplot(data=top10_waste, x="Country", y="Total Plastic Waste (MT)", palette="Reds_r")
plt.title("Top 6 Countries by Total Plastic Waste Production")
plt.xlabel("Country")
plt.ylabel("Total Plastic Waste (Million Tons)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Correlation Analysis

```python
import numpy as np
correlation_matrix = df[['Total Plastic Waste (MT)', 'Recycling Rate (%)', 'Per Capita Waste (KG)']].corr()
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm",
            mask=mask, linewidths=0.5, vmin=-1, vmax=1)
plt.title("Correlation Heatmap (Lower Triangle)")
plt.tight_layout()
plt.show()
```

Relationship Between Total Plastic Waste and Recycling Rate

```python
sns.lmplot(
    data=df,
    x="Recycling Rate (%)",
    y="Total Plastic Waste (MT)",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)
plt.title("Total Plastic Waste vs. Recycling Rate")
plt.xlabel("Recycling Rate (%)")
plt.ylabel("Total Plastic Waste (MT)")
plt.tight_layout()
plt.show()
```

```python
# In this chart, a positive correlation between Total Plastic Waste and Recycling Rate is observed.
# Countries with higher waste production tend to have slightly higher recycling rates.
# However, many high-waste countries still have low recycling rates, indicating room for improvement.
```

Top 20 and Bottom 20 Countries by Recycling Rate – Bar Chart

```python
df_sorted = df.sort_values(by='Recycling Rate (%)', ascending=False)
top_20_recycling = df_sorted.head(20)

plt.figure(figsize=(12, 8))
sns.barplot(data=top_20_recycling, x='Recycling Rate (%)', y='Country', palette='Greens_r')
plt.title('Top 20 Countries with Highest Recycling Rates')
plt.xlabel('Recycling Rate (%)')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
df_sorted_2 = df.sort_values(by='Recycling Rate (%)', ascending=True)
least_20_recycling = df_sorted_2.head(20)

plt.figure(figsize=(12, 8))
sns.barplot(data=least_20_recycling, x='Recycling Rate (%)', y='Country', palette='Reds_r')
plt.title('Bottom 20 Countries with Lowest Recycling Rates')
plt.xlabel('Recycling Rate (%)')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
top_20_recycling = df_merged[df_merged["Country"].isin(top_20_recycling["Country"])].copy()
least_20_recycling = df_merged[df_merged["Country"].isin(least_20_recycling["Country"])].copy()

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

top_20_sorted = top_20_recycling.sort_values("Recycling Rate (%)", ascending=False)
sns.barplot(
    data=top_20_sorted,
    x="Recycling Rate (%)",
    y="Country",
    hue="Development Level",
    dodge=False,
    palette={"Developed": "green", "Developing": "orange", "Underdeveloped": "red"},
    ax=axes[0]
)
axes[0].set_title("Top 20 Countries - Highest Recycling Rates")
axes[0].set_xlabel("Recycling Rate (%)")
axes[0].set_ylabel("Country")

least_20_sorted = least_20_recycling.sort_values("Recycling Rate (%)", ascending=True)
sns.barplot(
    data=least_20_sorted,
    x="Recycling Rate (%)",
    y="Country",
    hue="Development Level",
    dodge=False,
    palette={"Developed": "green", "Developing": "orange", "Underdeveloped": "red"},
    ax=axes[1]
)
axes[1].set_title("Bottom 20 Countries - Lowest Recycling Rates")
axes[1].set_xlabel("Recycling Rate (%)")
axes[1].set_ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# Looking at the charts above, the countries with the highest recycling rates are predominantly developed nations.
# Countries with the lowest recycling rates are mostly developing or underdeveloped countries.
```

Grouping by Coastal Waste Risk and Calculating Average Recycling Rates

```python
average_recycling_by_risk = df.groupby('Coastal Waste Risk')['Recycling Rate (%)'].mean()
print('Average Recycling Rate by Coastal Waste Risk:')
print(average_recycling_by_risk)
```

Recycling Rate and Coastal Waste Risk – Box Plot

```python
high_risk_df = df[df['Coastal Waste Risk'] == 'High']
low_risk_df = df[df['Coastal Waste Risk'] == 'Low']
medium_risk_df = df[df['Coastal Waste Risk'] == 'Medium']

data_to_plot = [high_risk_df['Recycling Rate (%)'].values,
                medium_risk_df['Recycling Rate (%)'].values,
                low_risk_df['Recycling Rate (%)'].values]

fig, ax = plt.subplots(figsize=(10, 6))
ax.boxplot(data_to_plot, labels=['High', 'Medium', 'Low'])
ax.set_title('Recycling Rate by Coastal Waste Risk Level')
ax.set_xlabel('Coastal Waste Risk Level')
ax.set_ylabel('Recycling Rate (%)')
plt.tight_layout()
plt.show()
```

Scatter Plot: Total Plastic Waste Production vs. Recycling Relationship

```python
df["Plastic Recycling Efficiency"] = df["Recycling Rate (%)"] / df["Per Capita Waste (KG)"]
top_efficient = df.sort_values(by="Plastic Recycling Efficiency", ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(data=top_efficient, x="Country", y="Plastic Recycling Efficiency", palette="viridis")
plt.title("Top 10 Countries by Plastic Recycling Efficiency")
plt.xlabel("Country")
plt.ylabel("Recycling Efficiency (Rate / Per Capita Waste)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Turkey's Position in Plastic Waste Production and Recycling

```python
turkey_data = df[df['Country'] == 'Turkey']
turkey_data2 = df_merged[df_merged['Country'] == 'Turkey']
country = turkey_data['Country'].iloc[0]
total_waste = turkey_data['Total Plastic Waste (MT)'].iloc[0]
sources = turkey_data['Main Source'].iloc[0]
recycling_rate = turkey_data['Recycling Rate (%)'].iloc[0]
per_capita_waste = turkey_data['Per Capita Waste (KG)'].iloc[0]
coastal_risk = turkey_data['Coastal Waste Risk'].iloc[0]
development_level = turkey_data2['Development Level'].iloc[0]
efficiency = turkey_data['Plastic Recycling Efficiency'].iloc[0]

print("Turkey's Data:")
print(f"Country: {country}")
print(f"Total Plastic Waste (Million Tons): {total_waste}")
print(f"Main Sources: {sources}")
print(f"Recycling Rate (%): {recycling_rate}")
print(f"Per Capita Waste (KG): {per_capita_waste}")
print(f"Coastal Waste Risk: {coastal_risk}")
print(f"Development Level: {development_level} Country")
print(f"Plastic Recycling Efficiency: {efficiency}")
```
